# View the processed loans

A quick look at `data/processed/loans_processed.parquet`, the model-ready table written by `run_pipeline.py`. Use the project `.venv` as the kernel.

```python
import pandas as pd

from credit_risk.config import LOANS_PROCESSED_PARQUET, TARGET

pd.set_option("display.max_columns", 100)
df = pd.read_parquet(LOANS_PROCESSED_PARQUET)
df.shape
```

## Rows and default rate per split

```python
df.groupby("split")[TARGET].agg(loans="size", default_rate="mean").round(4)
```

```python
df.head()
```

## Column types and missing values

```python
pd.DataFrame({"dtype": df.dtypes.astype(str), "missing_%": (df.isna().mean() * 100).round(1)})
```
